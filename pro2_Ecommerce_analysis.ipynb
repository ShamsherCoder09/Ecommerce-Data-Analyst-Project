```python
# Remoce duplicate records
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Replace invalid values
# Replace comman invalid values with NaN
df.replace(['N/A', 'NULL',''],np.nan,inplace=True)
df.sample(7)
```

```python
# step 7 remove extra spaces
df  = df.apply(lambda x : x.str.strip() if x.dtype == 'object' else x)
```

```python
# step 8 Standardize text format
# convert text into proper format
df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
```

```python
# step 9 valid email address
# keep only valid email address
df = df[df['Email'].str.contains('@',na=False)]
```

```python
# step 10 Convert Date Format
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce') # NAT
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors='coerce')
```

```python
# 11 convert numeric columns 
cols = ['Qty','Unit_Price','Discount']
for c in cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')
```

```python
# step 12 Remove invalid quantity
# reomve records having negative or zero quantity
df = df[df['Qty']>0]
```

```python
# step 13 fill missing values
df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# step 14 Feature Engineering
df['Sales'] = df['Qty']* df['Unit_Price']
df.head()
```

```python
# Net Amount 
df['Net_Amount']= (df['Sales'] - df['Discount']/10)
df.head()
```

```python
# Profit (Example 20%)
df['Profit'] = df['Net_Amount']*0.02
df.head()
```

```python
# order month
df['month']  = df['Order_Date'].dt.month_name()
```

```python
# order year
df['Year'] = df['Order_Date'].dt.year
df.head()
```

```python
df['Weekday'] = df['Order_Date'].dt.day_name()
df.head()
```

```python
# step 15 final visualization
df.info()
```

```python
# step 16 Export clean data
df.to_csv('Clean_Ecommerce_data.csv', index=False)
```
